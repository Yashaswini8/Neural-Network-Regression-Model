```python
import torch
import torch.nn as nn
import torch.optim as optim
import matplotlib.pyplot as plt
```

## Dataset

```python
# Sample Dataset
X = torch.tensor([
    [5, 10],
    [10, 15],
    [15, 20],
    [20, 25],
    [25, 30],
    [30, 35],
    [35, 40],
    [40, 45]
], dtype=torch.float32)

y = torch.tensor([
    [15.2],
    [24.8],
    [35.1],
    [44.9],
    [55.3],
    [64.7],
    [75.4],
    [84.8]
], dtype=torch.float32)
```

## Build Neural Network

```python
class NeuralNet(nn.Module):
    def __init__(self):
        super().__init__()
        self.fc1 = nn.Linear(2, 16)
        self.fc2 = nn.Linear(16, 8)
        self.fc3 = nn.Linear(8, 1)
        self.relu = nn.ReLU()

    def forward(self, x):
        x = self.relu(self.fc1(x))
        x = self.relu(self.fc2(x))
        x = self.fc3(x)
        return x
```

## Initialize Model, Loss & Optimizer

```python
ai_brain = NeuralNet()
criterion = nn.MSELoss()
optimizer = optim.Adam(ai_brain.parameters(), lr=0.01)
```

## Train Model

```python
losses = []
epochs = 500

for epoch in range(epochs):
    optimizer.zero_grad()
    outputs = ai_brain(X)
    loss = criterion(outputs, y)
    loss.backward()
    optimizer.step()
    losses.append(loss.item())

    if (epoch+1) % 100 == 0:
        print(f"Epoch {epoch+1}, Loss: {loss.item():.6f}")
```

## Plot Training Loss

```python
plt.plot(losses)
plt.title("Training Loss Vs Iteration")
plt.xlabel("Iterations")
plt.ylabel("Training Loss")
plt.show()
```

## Predict New Sample

```python
new_sample = torch.tensor([[35, 45]], dtype=torch.float32)
prediction = ai_brain(new_sample)
print("New Sample Input: [35, 45]")
print("Predicted Output:", round(prediction.item(), 5))
```
